```python
# ============================================================
# MINIMUM SPANNING TREE USING KRUSKAL'S AND PRIM'S ALGORITHMS
# ============================================================

# Graph:
# Vertices = 0, 1, 2, 3, 4, 5, 6
#
# Edges:
# (0, 1, 4)
# (0, 2, 3)
# (1, 2, 1)
# (1, 3, 2)
# (2, 3, 4)
# (2, 4, 2)
# (3, 5, 3)
# (4, 5, 5)
# (4, 6, 3)

vertices = 7

edges = [
    (0, 1, 4),
    (0, 2, 3),
    (1, 2, 1),
    (1, 3, 2),
    (2, 3, 4),
    (2, 4, 2),
    (3, 5, 3),
    (4, 5, 5),
    (4, 6, 3)
]


# ============================================================
# KRUSKAL'S ALGORITHM
# ============================================================

def find(parent, vertex):

    if parent[vertex] != vertex:
        parent[vertex] = find(parent, parent[vertex])

    return parent[vertex]


def union(parent, rank, u, v):

    root_u = find(parent, u)
    root_v = find(parent, v)

    if root_u != root_v:

        if rank[root_u] < rank[root_v]:
            parent[root_u] = root_v

        elif rank[root_u] > rank[root_v]:
            parent[root_v] = root_u

        else:
            parent[root_v] = root_u
            rank[root_u] += 1

        return True

    return False


def kruskal(vertices, edges):

    # Sort edges by weight
    sorted_edges = sorted(edges, key=lambda x: x[2])

    parent = list(range(vertices))
    rank = [0] * vertices

    mst = []
    total_weight = 0

    print("KRUSKAL'S ALGORITHM")
    print("-------------------")

    for u, v, weight in sorted_edges:

        if union(parent, rank, u, v):

            mst.append((u, v, weight))
            total_weight += weight

            print(
                f"Selected: {u} - {v}  Weight = {weight}"
            )

        if len(mst) == vertices - 1:
            break

    return mst, total_weight


# ============================================================
# PRIM'S ALGORITHM
# ============================================================

def prim(vertices, edges):

    # Create adjacency list
    graph = [[] for _ in range(vertices)]

    for u, v, weight in edges:

        graph[u].append((v, weight))
        graph[v].append((u, weight))

    visited = [False] * vertices

    mst = []
    total_weight = 0

    # Start from vertex 0
    visited[0] = True

    print("\nPRIM'S ALGORITHM")
    print("----------------")

    for _ in range(vertices - 1):

        minimum_edge = None
        minimum_weight = float('inf')

        # Find minimum weight edge connecting
        # visited and unvisited vertices
        for u in range(vertices):

            if visited[u]:

                for v, weight in graph[u]:

                    if not visited[v] and weight < minimum_weight:

                        minimum_weight = weight
                        minimum_edge = (u, v, weight)

        if minimum_edge is not None:

            u, v, weight = minimum_edge

            visited[v] = True

            mst.append((u, v, weight))
            total_weight += weight

            print(
                f"Selected: {u} - {v}  Weight = {weight}"
            )

    return mst, total_weight


# ============================================================
# RUN KRUSKAL
# ============================================================

kruskal_mst, kruskal_weight = kruskal(vertices, edges)


# ============================================================
# RUN PRIM
# ============================================================

prim_mst, prim_weight = prim(vertices, edges)


# ============================================================
# DISPLAY RESULTS
# ============================================================

print("\n==========================================")
print("FINAL MST RESULTS")
print("==========================================")

print("\nKruskal's MST:")
for u, v, weight in kruskal_mst:
    print(f"{u} - {v} : {weight}")

print("Total Minimum Weight =", kruskal_weight)


print("\nPrim's MST:")
for u, v, weight in prim_mst:
    print(f"{u} - {v} : {weight}")

print("Total Minimum Weight =", prim_weight)


# ============================================================
# VERIFY BOTH RESULTS
# ============================================================

print("\n==========================================")
print("VERIFICATION")
print("==========================================")

if kruskal_weight == prim_weight:
    print("Both algorithms produce the same MST weight.")
    print("Minimum Spanning Tree Weight =", kruskal_weight)
else:
    print("The MST weights are different.")


# ============================================================
# TIME COMPLEXITY
# ============================================================

print("\n==========================================")
print("TIME COMPLEXITY")
print("==========================================")

print("Kruskal's Algorithm : O(E log E)")
print("Prim's Algorithm    : O(V^2) using adjacency matrix/list search")
print()
print("For this graph:")
print("V = 7")
print("E = 9")
```
