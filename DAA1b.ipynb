```python
import matplotlib.pyplot as plt

# Binary Search
def binary_search(arr, key):
    low, high = 0, len(arr) - 1
    probes = 0

    while low <= high:
        probes += 1
        mid = (low + high) // 2

        if arr[mid] == key:
            return probes
        elif arr[mid] < key:
            low = mid + 1
        else:
            high = mid - 1

    return probes


# Interpolation Search
def interpolation_search(arr, key):
    low, high = 0, len(arr) - 1
    probes = 0

    while low <= high and key >= arr[low] and key <= arr[high]:

        probes += 1

        if low == high:
            return probes

        pos = low + ((key - arr[low]) * (high - low)) // \
                    (arr[high] - arr[low])

        if arr[pos] == key:
            return probes
        elif arr[pos] < key:
            low = pos + 1
        else:
            high = pos - 1

    return probes


# Roll numbers
roll_numbers = list(range(1, 10001))

# Test roll numbers
test_rolls = [1000, 2000, 3000, 4000, 5000,
              6000, 7000, 8000, 9000, 10000]

binary_probes = []
interpolation_probes = []

for roll in test_rolls:
    binary_probes.append(
        binary_search(roll_numbers, roll)
    )

    interpolation_probes.append(
        interpolation_search(roll_numbers, roll)
    )

# Plot Graph
plt.figure(figsize=(10, 6))

plt.plot(test_rolls,
         binary_probes,
         marker='o',
         label='Binary Search')

plt.plot(test_rolls,
         interpolation_probes,
         marker='s',
         label='Interpolation Search')

plt.title("Binary Search vs Interpolation Search")
plt.xlabel("Student Roll Number")
plt.ylabel("Number of Probes")
plt.legend()
plt.grid(True)

plt.show()
```
