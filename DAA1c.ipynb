```python
import random
import matplotlib.pyplot as plt

# Interpolation Search for floating point values
def interpolation_search(arr, key):

    low = 0
    high = len(arr) - 1
    comparisons = 0

    while low <= high and arr[low] <= key <= arr[high]:

        comparisons += 1

        if low == high:
            return comparisons

        pos = low + int(
            ((key - arr[low]) * (high - low))
            / (arr[high] - arr[low])
        )

        pos = max(low, min(pos, high))

        if abs(arr[pos] - key) < 1e-6:
            return comparisons

        elif arr[pos] < key:
            low = pos + 1

        else:
            high = pos - 1

    return comparisons


sizes = [10000, 50000, 100000]
avg_comparisons = []

for n in sizes:

    # Uniformly distributed float values
    arr = sorted(
        random.uniform(0.0, 1000.0)
        for _ in range(n)
    )

    total = 0
    trials = 100

    for _ in range(trials):

        key = random.choice(arr)

        total += interpolation_search(arr, key)

    avg = total / trials
    avg_comparisons.append(avg)

    print(f"Size={n}, Average Comparisons={avg:.2f}")

# Graph
plt.figure(figsize=(8,5))

plt.plot(
    sizes,
    avg_comparisons,
    marker='o',
    linewidth=2
)

plt.title(
    "Interpolation Search Performance"
)

plt.xlabel("Dataset Size")
plt.ylabel("Average Comparisons")
plt.grid(True)

plt.show()
```
